```python
import cv2
from ultralytics import YOLO

modelo = YOLO("results/model_yolo11n_detect_plates/weights/best.pt")

ruta_video = "./video_test.mp4"

captura = cv2.VideoCapture(ruta_video)

# Clases del modelo preentrenado:
# 0: persona, 1: bicicleta, 2: coche,
# 3: motocicleta, 5: autobús, 7: camión
clases_interes = [0, 1, 2, 3, 5, 7]

if not captura.isOpened():
    print("No se pudo abrir el vídeo.")
else:
    while True:
        correcto, fotograma = captura.read()

        if not correcto:
            break

        resultados = modelo.track(
            source=fotograma,
            persist=False,
            classes=clases_interes,
            conf=0.25,
            verbose=False
        )

        cajas = resultados[0].boxes

        print("Tracking:", cajas.is_track)
        print("IDs:", cajas.id)

        if cajas.id is not None:
            identificadores = cajas.id.int().cpu().tolist()
            clases = cajas.cls.int().cpu().tolist()

            for identificador, clase in zip(identificadores, clases):
                print(f"ID: {identificador} | Clase: {modelo.names[clase]}")

        # Dibujamos las cajas de detección
        fotograma_anotado = resultados[0].plot() 
        cv2.imshow("Deteccion de personas y vehiculos", fotograma_anotado)

        if cv2.waitKey(1) == 27:
            break

captura.release()
cv2.destroyAllWindows()
```
